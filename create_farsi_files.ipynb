```python
"""
مرحله 1: ترکیب دو فایل و حذف کلمات چندکلمه‌ای
- ورودی 1: Persian-Words-Database/distinct_words.txt (50000 خط اول)
- ورودی 2: corpora_farsi/farhang-farsi.tr (کل فایل)
- خروجی: corpora_farsi/corpora-farsi.tr 
- قوانین:
  ✅ نیم‌فاصله (U+200C) حفظ می‌شود
  ❌ کلمات با فاصله معمولی (U+0020) حذف می‌شوند
  ✅ حذف تکراری‌ها
"""

import os

# ===== تنظیمات =====
INPUT_FILE_1 = '/home/GAN_Writing_Farsi/Persian-Words-Database/distinct_words.txt'
INPUT_FILE_2 = '/home/GAN_Writing_Farsi/corpora_farsi/farhang-farsi.tr'
OUTPUT_FILE = '/home/GAN_Writing_Farsi/corpora_farsi/corpora-farsi.tr'
MAX_LINES_FILE1 = 50000

SPACE = ' '  # U+0020
ZWNJ = '\u200c'  # نیم‌فاصله

print("="*70)
print("🔧 مرحله 1: ترکیب و حذف کلمات چندکلمه‌ای")
print("="*70)
print(f"📥 ورودی 1: {INPUT_FILE_1} (50000 خط اول)")
print(f"📥 ورودی 2: {INPUT_FILE_2} (کل)")
print(f"📤 خروجی: {OUTPUT_FILE}\n")

# ===== تابع پردازش فایل =====
def process_file(filepath, max_lines=None):
    """
    خواندن و فیلتر کردن یک فایل
    """
    if not os.path.exists(filepath):
        print(f"⚠️ فایل {filepath} یافت نشد - رد می‌شود")
        return set(), {'total': 0, 'empty': 0, 'has_space': 0, 'has_zwnj': 0, 'valid': 0}
    
    print(f"📖 در حال خواندن: {filepath}")
    
    with open(filepath, 'r', encoding='utf-8') as f:
        lines = f.readlines()
    
    total_read = len(lines)
    
    if max_lines:
        lines = lines[:max_lines]
        print(f"   ✅ کل خطوط: {total_read:,} → محدود به: {len(lines):,}")
    else:
        print(f"   ✅ کل خطوط: {total_read:,}")
    
    valid_words = set()
    stats = {
        'total': len(lines),
        'empty': 0,
        'has_space': 0,
        'has_zwnj': 0,
        'valid': 0
    }
    
    for i, line in enumerate(lines, 1):
        word = line.strip()
        
        # ❌ خط خالی
        if not word:
            stats['empty'] += 1
            continue
        
        # ❌ دارای فاصله معمولی
        if SPACE in word:
            stats['has_space'] += 1
            if i <= 10:  # نمایش 10 نمونه اول
                print(f"      {i:6d}. ❌ '{word}' → دارای فاصله")
            continue
        
        # ✅ معتبر
        if ZWNJ in word:
            stats['has_zwnj'] += 1
        
        valid_words.add(word)
        stats['valid'] += 1
        
        # نمایش نمونه
        if i <= 10:
            has_zwnj_mark = '🔸' if ZWNJ in word else '  '
            print(f"      {i:6d}. ✅ '{word}' {has_zwnj_mark}")
    
    return valid_words, stats

# ===== پردازش فایل اول =====
print("\n" + "="*70)
print("📁 پردازش فایل 1: distinct_words.txt")
print("="*70)
words_set_1, stats_1 = process_file(INPUT_FILE_1, MAX_LINES_FILE1)

# ===== پردازش فایل دوم =====
print("\n" + "="*70)
print("📁 پردازش فایل 2: corpora-farsi.tr (موجود)")
print("="*70)
words_set_2, stats_2 = process_file(INPUT_FILE_2)

# ===== ترکیب =====
print("\n" + "="*70)
print("🔗 ترکیب و حذف تکراری‌ها")
print("="*70)
print(f"   📦 فایل 1: {len(words_set_1):,} کلمه یونیک")
print(f"   📦 فایل 2: {len(words_set_2):,} کلمه یونیک")

all_words = words_set_1 | words_set_2  # Union
duplicates = len(words_set_1) + len(words_set_2) - len(all_words)

print(f"   ➕ مجموع:  {len(words_set_1) + len(words_set_2):,} کلمه")
print(f"   ♻️ تکراری: {duplicates:,} کلمه")
print(f"   ✅ یونیک:   {len(all_words):,} کلمه")

# ===== مرتب‌سازی =====
sorted_words = sorted(all_words)
print(f"\n   🔤 مرتب‌سازی الفبایی انجام شد")

# ===== ذخیره =====
print("\n💾 در حال ذخیره...")

# ایجاد پوشه
output_dir = os.path.dirname(OUTPUT_FILE)
if not os.path.exists(output_dir):
    os.makedirs(output_dir)
    print(f"   ✅ پوشه {output_dir} ایجاد شد")

with open(OUTPUT_FILE, 'w', encoding='utf-8') as f:
    f.write('\n'.join(sorted_words))

print(f"   ✅ {len(sorted_words):,} کلمه ذخیره شد")

# ===== آمار کلی =====
total_input = stats_1['total'] + stats_2['total']
total_empty = stats_1['empty'] + stats_2['empty']
total_space = stats_1['has_space'] + stats_2['has_space']
total_zwnj = stats_1['has_zwnj'] + stats_2['has_zwnj']

words_with_zwnj = [w for w in all_words if ZWNJ in w]

print("\n" + "="*70)
print("📊 آمار نهایی:")
print("="*70)
print(f"   📥 کل خطوط پردازش شده:     {total_input:,}")
print(f"      ├─ فایل 1:                {stats_1['total']:,}")
print(f"      └─ فایل 2:                {stats_2['total']:,}")
print(f"   ❌ خط خالی:                  {total_empty:,}")
print(f"   ❌ دارای فاصله معمولی:       {total_space:,} ({100*total_space/total_input if total_input else 0:.1f}%)")
print(f"   ✅ کلمات یونیک نهایی:       {len(all_words):,}")
print(f"   🔸 با نیم‌فاصله:            {len(words_with_zwnj):,} ({100*len(words_with_zwnj)/len(all_words):.1f}%)")
print("="*70)
print(f"📁 فایل خروجی: {OUTPUT_FILE}")
print("="*70)
print("✅ مرحله 1 تمام شد!\n")

# ===== نمایش نمونه =====
print("🔍 نمونه کلمات نهایی (10 مورد اول):")
print("="*70)
for i, word in enumerate(sorted_words[:10], 1):
    has_zwnj_mark = '🔸' if ZWNJ in word else '  '
    print(f"   {i:2d}. '{word}' {has_zwnj_mark}")
print("="*70)
```

```python
"""
مرحله 2: فیلتر کلمات بر اساس الفبای مجاز و طول مرئی
- ورودی: corpora_farsi/corpora-farsi.tr
- خروجی: corpora_farsi/corpora-farsi-myChars.tr
- الفبای مجاز: 30 کاراکتر (نیم‌فاصله + آابتثجحخدذرزسشصضعفقلمنهوپچکگی)
- حداکثر طول مرئی: 10 حرف (بدون نیم‌فاصله)
"""

import os

# ===== تنظیمات =====
INPUT_FILE = '/home/GAN_Writing_Farsi/corpora_farsi/corpora-farsi.tr'
OUTPUT_FILE = '/home/GAN_Writing_Farsi/corpora_farsi/corpora-farsi-myChars.tr'

# ⚠️ فقط نیم‌فاصله - بدون فاصله معمولی
ALLOWED_CHARS = set('آابتثجحخدذرزسشصضعفقلمنهوپچکگی\u200c')  # 30 کاراکتر
MAX_CHARS = 10  # حداکثر طول مرئی (بدون نیم‌فاصله)

def count_visible_chars(word):
    """شمارش حروف بدون نیم‌فاصله"""
    return len(word.replace('\u200c', ''))

print("="*70)
print("🔧 مرحله 2: فیلتر بر اساس الفبای مجاز و طول مرئی")
print("="*70)
print(f"📥 ورودی: {INPUT_FILE}")
print(f"📤 خروجی: {OUTPUT_FILE}")
print(f"🔤 الفبای مجاز: {len(ALLOWED_CHARS)} کاراکتر")
print(f"📏 حداکثر طول مرئی: {MAX_CHARS} حرف (بدون نیم‌فاصله)")

# نمایش کاراکترها
chars_display = list(ALLOWED_CHARS)
chars_display.sort()
zwnj_index = chars_display.index('\u200c') if '\u200c' in chars_display else -1
if zwnj_index >= 0:
    chars_display[zwnj_index] = 'نیم‌فاصله'
print(f"   → {' '.join(chars_display)}\n")

# ===== بررسی وجود فایل ورودی =====
if not os.path.exists(INPUT_FILE):
    print(f"❌ خطا: فایل {INPUT_FILE} پیدا نشد!")
    print("⚠️ لطفاً ابتدا step1_remove_multiword.py را اجرا کنید")
    exit(1)

# ===== خواندن فایل ورودی =====
print("📖 در حال خواندن فایل ورودی...")
with open(INPUT_FILE, 'r', encoding='utf-8') as f:
    input_words = f.read().strip().split('\n')

print(f"✅ تعداد کلمات ورودی: {len(input_words):,}\n")

# ===== فیلتر کردن =====
print("🔍 در حال فیلتر کردن...\n")

valid_words = []
invalid_samples = []
stats = {
    'total': len(input_words),
    'invalid_chars': 0,
    'too_long': 0,
    'too_short': 0,  # ✅ اضافه شد
    'valid': 0
}

for i, word in enumerate(input_words, 1):
    word = word.strip()
    
    if not word:
        continue
    
    # ✅ چک طول مرئی (اول این را چک کنیم)
    visible_len = count_visible_chars(word)
    
    # حذف کلمات خیلی کوتاه (فقط نیم‌فاصله)
    if visible_len == 0:
        stats['too_short'] += 1
        if i <= 20 or (i % 5000 == 0):
            print(f"   {i:6d}. ❌ '{word}' → طول مرئی=0 (فقط نیم‌فاصله)")
        continue
    
    if visible_len > MAX_CHARS:
        stats['too_long'] += 1
        
        if len(invalid_samples) < 20:
            invalid_samples.append({
                'index': i,
                'word': word,
                'invalid': [f'طول_مرئی={visible_len}>{MAX_CHARS}']
            })
        
        if i <= 20 or (i % 5000 == 0):
            print(f"   {i:6d}. ❌ '{word}' → طول مرئی={visible_len} (کل={len(word)})")
        
        continue
    
    # چک کاراکترهای نامعتبر
    invalid_chars = []
    for char in word:
        if char not in ALLOWED_CHARS:
            char_display = f"{char}(U+{ord(char):04X})"
            # برای فاصله معمولی پیام خاص
            if char == ' ':
                char_display = "SPACE(U+0020)"
            invalid_chars.append(char_display)
    
    if invalid_chars:
        stats['invalid_chars'] += 1
        
        if len(invalid_samples) < 40:  # بیشتر کنیم برای دو نوع خطا
            invalid_samples.append({
                'index': i,
                'word': word,
                'invalid': invalid_chars
            })
        
        if i <= 20 or (i % 5000 == 0):
            print(f"   {i:6d}. ❌ '{word}' → {', '.join(invalid_chars[:3])}")
        
        continue
    
    # ✅ معتبر
    valid_words.append(word)
    stats['valid'] += 1
    
    if i <= 20 or (i % 5000 == 0):
        has_zwnj = '🔸' if '\u200c' in word else '  '
        print(f"   {i:6d}. ✅ '{word}' {has_zwnj} (طول مرئی={visible_len})")

# ===== ذخیره =====
print("\n💾 در حال ذخیره...")
with open(OUTPUT_FILE, 'w', encoding='utf-8') as f:
    f.write('\n'.join(valid_words))

print(f"✅ {len(valid_words):,} کلمه ذخیره شد\n")

# ===== آمار نهایی =====
lengths_total = [len(w) for w in valid_words]
lengths_visible = [count_visible_chars(w) for w in valid_words]
with_zwnj = [w for w in valid_words if '\u200c' in w]

print("="*70)
print("📊 آمار نهایی:")
print("="*70)
print(f"   📥 کل کلمات ورودی:         {stats['total']:,}")
print(f"   ❌ کاراکتر نامعتبر:         {stats['invalid_chars']:,} ({100*stats['invalid_chars']/stats['total']:.1f}%)")
print(f"   ❌ طول بیش از حد:           {stats['too_long']:,} ({100*stats['too_long']/stats['total']:.1f}%)")
print(f"   ❌ طول خیلی کم (=0):        {stats['too_short']:,} ({100*stats['too_short']/stats['total']:.2f}%)")
print(f"   ✅ معتبر (الفبای مجاز):    {stats['valid']:,} ({100*stats['valid']/stats['total']:.1f}%)")
print(f"   🔸 با نیم‌فاصله:            {len(with_zwnj):,} ({100*len(with_zwnj)/len(valid_words):.1f}%)")
print()
print(f"   📏 طول کل (با نیم‌فاصله):")
if lengths_total:
    print(f"      • کوتاه‌ترین:           {min(lengths_total)} حرف")
    print(f"      • بلندترین:             {max(lengths_total)} حرف")
    print(f"      • متوسط:                {sum(lengths_total)/len(lengths_total):.2f} حرف")
print()
print(f"   📐 طول مرئی (بدون نیم‌فاصله):")
if lengths_visible:
    print(f"      • کوتاه‌ترین:           {min(lengths_visible)} حرف")
    print(f"      • بلندترین:             {max(lengths_visible)} حرف")
    print(f"      • متوسط:                {sum(lengths_visible)/len(lengths_visible):.2f} حرف")
print("="*70)
print(f"📁 فایل خروجی: {OUTPUT_FILE}")
print("="*70)
print("✅ مرحله 2 تمام شد!\n")

# نمایش نمونه کلمات حذف شده
if invalid_samples:
    print("\n🔍 نمونه کلمات حذف شده (20 مورد اول):")
    print("="*70)
    
    # جدا کردن دو نوع خطا
    too_long_samples = [s for s in invalid_samples if any('طول_مرئی' in str(inv) for inv in s['invalid'])]
    invalid_char_samples = [s for s in invalid_samples if s not in too_long_samples]
    
    if too_long_samples:
        print("\n   📏 طول بیش از حد:")
        for sample in too_long_samples[:10]:
            vis_len = count_visible_chars(sample['word'])
            print(f"      {sample['index']:6d}. '{sample['word']}'")
            print(f"              → طول کل={len(sample['word'])}, مرئی={vis_len} > {MAX_CHARS}")
    
    if invalid_char_samples:
        print("\n   🔤 کاراکتر نامعتبر:")
        for sample in invalid_char_samples[:10]:
            print(f"      {sample['index']:6d}. '{sample['word']}'")
            print(f"              → {', '.join(sample['invalid'][:5])}")
    
    print("="*70)
```

```python
import os
import torch.utils.data as D
import random
import cv2
import numpy as np
from pairs_idx_wid_ABAN import wid2label_tr, wid2label_te

CREATE_PAIRS = False

IMG_HEIGHT = 64
IMG_WIDTH = 216
MAX_CHARS = 10
NUM_CHANNEL = 15
# NUM_CHANNEL = 50
EXTRA_CHANNEL = NUM_CHANNEL + 1
NUM_WRITERS = 500  # ABAN: 350 train + 150 test
NORMAL = True
OUTPUT_MAX_LEN = MAX_CHARS + 2  # <GO>+groundtruth+<END>

"""مسیر تصاویر کلمات ABAN"""
img_base = "./datasets/aban/words"
text_corpus = "./corpora_farsi/corpora-farsi-myChars.tr"

with open(text_corpus, "r", encoding='utf-8') as _f:
    text_corpus = _f.read().strip().split('\n')

src = "Groundtruth_farsi/gan.aban.tr_va.gt.filter27"
tar = "Groundtruth_farsi/gan.aban.test.gt.filter27"


def labelDictionary():
    """
    ساخت دیکشنری الفبای فارسی
    30 کاراکتر: آابتثجحخدذرزسشصضعفقلمنهوپچکگی + نیم‌فاصله
    """
    ZWNJ = '\u200c'  # Zero-Width Non-Joiner
    labels = list('آابتثجحخدذرزسشصضعفقلمنهوپچکگی' + ZWNJ)
    letter2index = {label: n for n, label in enumerate(labels)}
    index2letter = {v: k for k, v in letter2index.items()}
    return len(labels), letter2index, index2letter


num_classes, letter2index, index2letter = labelDictionary()
tokens = {"GO_TOKEN": 0, "END_TOKEN": 1, "PAD_TOKEN": 2}
num_tokens = len(tokens.keys())
vocab_size = num_classes + num_tokens


def edits1(word, min_len=2, max_len=MAX_CHARS):
    """
    ایجاد یک ویرایش (edit distance 1) از کلمه فارسی
    عملیات: حذف، جابجایی، جایگزینی، درج
    توجه: طول بر اساس حروف مرئی (بدون نیم‌فاصله) محاسبه می‌شود
    """
    ZWNJ = '\u200c'
    letters = list('آابتثجحخدذرزسشصضعفقلمنهوپچکگی' + ZWNJ)
    
    splits = [(word[:i], word[i:]) for i in range(len(word) + 1)]
    deletes = [L + R[1:] for L, R in splits if R]
    transposes = [L + R[1] + R[0] + R[2:] for L, R in splits if len(R) > 1]
    replaces = [L + c + R[1:] for L, R in splits if R for c in letters]
    inserts = [L + c + R for L, R in splits for c in letters]
    
    # محاسبه طول مرئی (بدون نیم‌فاصله)
    visible_len = len([c for c in word if c != ZWNJ])
    
    if visible_len <= min_len:
        # کلمه کوتاه: فقط جابجایی، جایگزینی، درج
        candidates = list(set(transposes + replaces + inserts))
    elif visible_len >= max_len:
        # کلمه بلند: فقط حذف، جابجایی، جایگزینی
        candidates = list(set(deletes + transposes + replaces))
    else:
        # کلمه متوسط: همه عملیات
        candidates = list(set(deletes + transposes + replaces + inserts))
    
    if not candidates:
        return word
    return random.choice(candidates)


class ABAN_words(D.Dataset):
    def __init__(self, data_dict, oov):
        self.data_dict = data_dict
        self.oov = oov
        self.output_max_len = OUTPUT_MAX_LEN

    def new_ed1(self, word_ori):
        """
        ایجاد یک کلمه جدید با edit distance 1 از کلمه اصلی
        word_ori مثال: [0, 15, 27, 13, 32, 31, 1, 2, 2, 2]
        """
        word = word_ori.copy()
        start = word.index(tokens["GO_TOKEN"])
        fin = word.index(tokens["END_TOKEN"])
        word = "".join([index2letter[i - num_tokens] for i in word[start + 1 : fin]])
        new_word = edits1(word)
        label = np.array(self.label_padding(new_word, num_tokens))
        return label

    def __getitem__(self, wid_idx_num):
        words = self.data_dict[wid_idx_num]
        """shuffle images"""
        np.random.shuffle(words)

        wids = list()
        idxs = list()
        imgs = list()
        img_widths = list()
        labels = list()

        for word in words:
            wid, idx = word[0].split(",")
            img, img_width = self.read_image_single(idx)
            label = self.label_padding(" ".join(word[1:]), num_tokens)
            wids.append(wid)
            idxs.append(idx)
            imgs.append(img)
            img_widths.append(img_width)
            labels.append(label)

        if len(list(set(wids))) != 1:
            print("Error! writer id differs")
            exit()

        final_wid = wid_idx_num
        num_imgs = len(imgs)
        if num_imgs >= EXTRA_CHANNEL:
            final_img = np.stack(imgs[:EXTRA_CHANNEL], axis=0)  # 64, h, w
            final_idx = idxs[:EXTRA_CHANNEL]
            final_img_width = img_widths[:EXTRA_CHANNEL]
            final_label = labels[:EXTRA_CHANNEL]
        else:
            final_idx = idxs
            final_img = imgs
            final_img_width = img_widths
            final_label = labels

            while len(final_img) < EXTRA_CHANNEL:
                num_cp = EXTRA_CHANNEL - len(final_img)
                final_idx = final_idx + idxs[:num_cp]
                final_img = final_img + imgs[:num_cp]
                final_img_width = final_img_width + img_widths[:num_cp]
                final_label = final_label + labels[:num_cp]
            final_img = np.stack(final_img, axis=0)

        _id = np.random.randint(EXTRA_CHANNEL)
        img_xt = final_img[_id : _id + 1]
        if self.oov:
            label_xt = np.random.choice(text_corpus)
            label_xt = np.array(self.label_padding(label_xt, num_tokens))
            label_xt_swap = np.random.choice(text_corpus)
            label_xt_swap = np.array(self.label_padding(label_xt_swap, num_tokens))
        else:
            label_xt = final_label[_id]
            label_xt_swap = self.new_ed1(label_xt)

        final_idx = np.delete(final_idx, _id, axis=0)
        final_img = np.delete(final_img, _id, axis=0)
        final_img_width = np.delete(final_img_width, _id, axis=0)
        final_label = np.delete(final_label, _id, axis=0)

        return (
            "src",
            final_wid,
            final_idx,
            final_img,
            final_img_width,
            final_label,
            img_xt,
            label_xt,
            label_xt_swap,
        )

    def __len__(self):
        return len(self.data_dict)

    def read_image_single(self, file_name):
        """
        خواندن تصویر کلمه فارسی
        فرمت: .png (مشابه IAM)
        """
        url = os.path.join(img_base, file_name + ".png")
        img = cv2.imread(url, 0)

        if img is None and os.path.exists(url):
            # image is present but corrupted
            return np.zeros((IMG_HEIGHT, IMG_WIDTH)), 0

        rate = float(IMG_HEIGHT) / img.shape[0]
        img = cv2.resize(
            img,
            (int(img.shape[1] * rate) + 1, IMG_HEIGHT),
            interpolation=cv2.INTER_CUBIC,
        )
        img = img / 255.0  # 0-255 -> 0-1

        img = 1.0 - img
        img_width = img.shape[-1]

        if img_width > IMG_WIDTH:
            outImg = img[:, :IMG_WIDTH]
            img_width = IMG_WIDTH
        else:
            outImg = np.zeros((IMG_HEIGHT, IMG_WIDTH), dtype="float32")
            outImg[:, :img_width] = img
        outImg = outImg.astype("float32")

        mean = 0.5
        std = 0.5
        outImgFinal = (outImg - mean) / std
        return outImgFinal, img_width

    def label_padding(self, labels, num_tokens):
        """
        تبدیل رشته به لیست اندیس‌ها با padding
        مثال: "س ل ا م" -> [0, 15, 27, 13, 32, 1, 2, 2, 2, 2, 2, 2]
        """
        new_label_len = []
        ll = [letter2index[i] for i in labels]
        new_label_len.append(len(ll) + 2)
        ll = np.array(ll) + num_tokens
        ll = list(ll)
        ll = [tokens["GO_TOKEN"]] + ll + [tokens["END_TOKEN"]]
        num = self.output_max_len - len(ll)
        if not num == 0:
            ll.extend([tokens["PAD_TOKEN"]] * num)  # replace PAD_TOKEN
        return ll


def loadData(oov):
    """
    بارگذاری داده‌های آموزش و تست ABAN
    """
    gt_tr = src
    gt_te = tar

    with open(gt_tr, "r", encoding='utf-8') as f_tr:
        data_tr = f_tr.readlines()
        data_tr = [i.strip().split(" ") for i in data_tr]
        tr_dict = dict()
        for i in data_tr:
            wid = i[0].split(",")[0]
            if wid not in tr_dict.keys():
                tr_dict[wid] = [i]
            else:
                tr_dict[wid].append(i)
        new_tr_dict = dict()
        if CREATE_PAIRS:
            create_pairs(tr_dict)
        for k in tr_dict.keys():
            new_tr_dict[wid2label_tr[k]] = tr_dict[k]

    with open(gt_te, "r", encoding='utf-8') as f_te:
        data_te = f_te.readlines()
        data_te = [i.strip().split(" ") for i in data_te]
        te_dict = dict()
        for i in data_te:
            wid = i[0].split(",")[0]
            if wid not in te_dict.keys():
                te_dict[wid] = [i]
            else:
                te_dict[wid].append(i)
        new_te_dict = dict()
        if CREATE_PAIRS:
            create_pairs(te_dict)
        for k in te_dict.keys():
            new_te_dict[wid2label_te[k]] = te_dict[k]

    data_train = ABAN_words(new_tr_dict, oov)
    data_test = ABAN_words(new_te_dict, oov)
    return data_train, data_test


def create_pairs(ddict):
    """
    ایجاد جفت‌های (label, writer_id)
    """
    num = len(ddict.keys())
    label2wid = list(zip(range(num), ddict.keys()))
    print(label2wid)


if __name__ == "__main__":
    # تست سریع
    ZWNJ = '\u200c'
    print("=" * 60)
    print("📚 تست load_data_farsi.py")
    print("=" * 60)
    print(f"✓ تعداد کلاس‌ها (حروف فارسی): {num_classes}")
    print(f"✓ vocab_size (حروف + توکن‌ها): {vocab_size}")
    print(f"✓ NUM_WRITERS: {NUM_WRITERS}")
    print(f"✓ OUTPUT_MAX_LEN: {OUTPUT_MAX_LEN}")
    print("\n📖 نمونه الفبا:")
    for i, (char, idx) in enumerate(list(letter2index.items())[:5]):
        print(f"   '{char}' -> {idx}")
    print(f"   ...")
    zwnj_idx = letter2index[ZWNJ]
    print(f"   نیم‌فاصله (U+200C) -> {zwnj_idx}")
    
    print("\n🔄 تست edits1:")
    test_word = "سلام"
    for _ in range(3):
        edited = edits1(test_word)
        print(f"   '{test_word}' -> '{edited}'")
    
    print("=" * 60)
```

```python
import os
import torch.utils.data as D
import random
import string
import cv2
import numpy as np
from pairs_idx_wid_iam import wid2label_tr, wid2label_te

CREATE_PAIRS = False

IMG_HEIGHT = 64
IMG_WIDTH = 216
MAX_CHARS = 10
NUM_CHANNEL = 15
# NUM_CHANNEL = 50
EXTRA_CHANNEL = NUM_CHANNEL + 1
NUM_WRITERS = 500  # iam
NORMAL = True
OUTPUT_MAX_LEN = MAX_CHARS + 2  # <GO>+groundtruth+<END>

"""The folder of IAM word images, please change to your own one before run it!!"""
img_base = "./datasets/iam/words"
text_corpus = "corpora_english/brown-azAZ.tr"

with open(text_corpus, "r") as _f:
    text_corpus = _f.read().split()

src = "Groundtruth/gan.iam.tr_va.gt.filter27"
tar = "Groundtruth/gan.iam.test.gt.filter27"


def labelDictionary():
    labels = list(string.ascii_lowercase + string.ascii_uppercase)
    letter2index = {label: n for n, label in enumerate(labels)}
    index2letter = {v: k for k, v in letter2index.items()}
    return len(labels), letter2index, index2letter


num_classes, letter2index, index2letter = labelDictionary()
tokens = {"GO_TOKEN": 0, "END_TOKEN": 1, "PAD_TOKEN": 2}
num_tokens = len(tokens.keys())
vocab_size = num_classes + num_tokens


def edits1(word, min_len=2, max_len=MAX_CHARS):
    "All edits that are one edit away from `word`."
    letters = list(string.ascii_lowercase)
    splits = [(word[:i], word[i:]) for i in range(len(word) + 1)]
    deletes = [L + R[1:] for L, R in splits if R]
    transposes = [L + R[1] + R[0] + R[2:] for L, R in splits if len(R) > 1]
    replaces = [L + c + R[1:] for L, R in splits if R for c in letters]
    inserts = [L + c + R for L, R in splits for c in letters]
    if len(word) <= min_len:
        return random.choice(list(set(transposes + replaces + inserts)))
    elif len(word) >= max_len:
        return random.choice(list(set(deletes + transposes + replaces)))
    else:
        return random.choice(list(set(deletes + transposes + replaces + inserts)))


class IAM_words(D.Dataset):
    def __init__(self, data_dict, oov):
        self.data_dict = data_dict
        self.oov = oov
        self.output_max_len = OUTPUT_MAX_LEN

    # word [0, 15, 27, 13, 32, 31, 1, 2, 2, 2]
    def new_ed1(self, word_ori):
        word = word_ori.copy()
        start = word.index(tokens["GO_TOKEN"])
        fin = word.index(tokens["END_TOKEN"])
        word = "".join([index2letter[i - num_tokens] for i in word[start + 1 : fin]])
        new_word = edits1(word)
        label = np.array(self.label_padding(new_word, num_tokens))
        return label

    def __getitem__(self, wid_idx_num):
        # print("wid_idx_num: ", wid_idx_num)
        words = self.data_dict[wid_idx_num]
        """shuffle images"""
        np.random.shuffle(words)

        wids = list()
        idxs = list()
        imgs = list()
        img_widths = list()
        labels = list()

        for word in words:
            wid, idx = word[0].split(",")
            img, img_width = self.read_image_single(idx)
            label = self.label_padding(" ".join(word[1:]), num_tokens)
            wids.append(wid)
            idxs.append(idx)
            imgs.append(img)
            img_widths.append(img_width)
            labels.append(label)

        if len(list(set(wids))) != 1:
            print("Error! writer id differs")
            exit()

        final_wid = wid_idx_num
        num_imgs = len(imgs)
        if num_imgs >= EXTRA_CHANNEL:
            final_img = np.stack(imgs[:EXTRA_CHANNEL], axis=0)  # 64, h, w
            final_idx = idxs[:EXTRA_CHANNEL]
            final_img_width = img_widths[:EXTRA_CHANNEL]
            final_label = labels[:EXTRA_CHANNEL]
        else:
            final_idx = idxs
            final_img = imgs
            final_img_width = img_widths
            final_label = labels

            while len(final_img) < EXTRA_CHANNEL:
                num_cp = EXTRA_CHANNEL - len(final_img)
                final_idx = final_idx + idxs[:num_cp]
                final_img = final_img + imgs[:num_cp]
                final_img_width = final_img_width + img_widths[:num_cp]
                final_label = final_label + labels[:num_cp]
            final_img = np.stack(final_img, axis=0)

        _id = np.random.randint(EXTRA_CHANNEL)
        img_xt = final_img[_id : _id + 1]
        if self.oov:
            label_xt = np.random.choice(text_corpus)
            label_xt = np.array(self.label_padding(label_xt, num_tokens))
            label_xt_swap = np.random.choice(text_corpus)
            label_xt_swap = np.array(self.label_padding(label_xt_swap, num_tokens))
        else:
            label_xt = final_label[_id]
            label_xt_swap = self.new_ed1(label_xt)

        final_idx = np.delete(final_idx, _id, axis=0)
        final_img = np.delete(final_img, _id, axis=0)
        final_img_width = np.delete(final_img_width, _id, axis=0)
        final_label = np.delete(final_label, _id, axis=0)

        return (
            "src",
            final_wid,
            final_idx,
            final_img,
            final_img_width,
            final_label,
            img_xt,
            label_xt,
            label_xt_swap,
        )

    def __len__(self):
        return len(self.data_dict)

    def read_image_single(self, file_name):
        url = os.path.join(img_base, file_name + ".png")
        img = cv2.imread(url, 0)

        if img is None and os.path.exists(url):
            # image is present but corrupted
            return np.zeros((IMG_HEIGHT, IMG_WIDTH)), 0

        rate = float(IMG_HEIGHT) / img.shape[0]
        img = cv2.resize(
            img,
            (int(img.shape[1] * rate) + 1, IMG_HEIGHT),
            interpolation=cv2.INTER_CUBIC,
        )
        img = img / 255.0  # 0-255 -> 0-1

        img = 1.0 - img
        img_width = img.shape[-1]

        if img_width > IMG_WIDTH:
            outImg = img[:, :IMG_WIDTH]
            img_width = IMG_WIDTH
        else:
            outImg = np.zeros((IMG_HEIGHT, IMG_WIDTH), dtype="float32")
            outImg[:, :img_width] = img
        outImg = outImg.astype("float32")

        mean = 0.5
        std = 0.5
        outImgFinal = (outImg - mean) / std
        return outImgFinal, img_width

    def label_padding(self, labels, num_tokens):
        new_label_len = []
        ll = [letter2index[i] for i in labels]
        new_label_len.append(len(ll) + 2)
        ll = np.array(ll) + num_tokens
        ll = list(ll)
        ll = [tokens["GO_TOKEN"]] + ll + [tokens["END_TOKEN"]]
        num = self.output_max_len - len(ll)
        if not num == 0:
            ll.extend([tokens["PAD_TOKEN"]] * num)  # replace PAD_TOKEN
        return ll


def loadData(oov):
    gt_tr = src
    gt_te = tar

    with open(gt_tr, "r") as f_tr:
        data_tr = f_tr.readlines()
        data_tr = [i.strip().split(" ") for i in data_tr]
        tr_dict = dict()
        for i in data_tr:
            wid = i[0].split(",")[0]
            if wid not in tr_dict.keys():
                tr_dict[wid] = [i]
            else:
                tr_dict[wid].append(i)
        new_tr_dict = dict()
        if CREATE_PAIRS:
            create_pairs(tr_dict)
        for k in tr_dict.keys():
            new_tr_dict[wid2label_tr[k]] = tr_dict[k]

    with open(gt_te, "r") as f_te:
        data_te = f_te.readlines()
        data_te = [i.strip().split(" ") for i in data_te]
        te_dict = dict()
        for i in data_te:
            wid = i[0].split(",")[0]
            if wid not in te_dict.keys():
                te_dict[wid] = [i]
            else:
                te_dict[wid].append(i)
        new_te_dict = dict()
        if CREATE_PAIRS:
            create_pairs(te_dict)
        for k in te_dict.keys():
            new_te_dict[wid2label_te[k]] = te_dict[k]
    
    # print(new_te_dict)
    data_train = IAM_words(new_tr_dict, oov)
    data_test = IAM_words(new_te_dict, oov)
    return data_train, data_test


def create_pairs(ddict):
    num = len(ddict.keys())
    label2wid = list(zip(range(num), ddict.keys()))
    print(label2wid)


if __name__ == "__main__":
    # تست سریع
    print("=" * 60)
    print("📚 تست load_data.py")
    print("=" * 60)
    print(f"✓ تعداد کلاس‌ها (حروف انگلیسی): {num_classes}")
    print(f"✓ vocab_size (حروف + توکن‌ها): {vocab_size}")
    print(f"✓ NUM_WRITERS: {NUM_WRITERS}")
    print(f"✓ OUTPUT_MAX_LEN: {OUTPUT_MAX_LEN}")
    print("\n📖 نمونه الفبا:")
    for i, (char, idx) in enumerate(list(letter2index.items())[:5]):
        print(f"   '{char}' -> {idx}")
    print(f"   ...")
    # print(f"   نیم‌فاصله (\\u200c) -> {letter2index['\u200c']}")
    
    print("\n🔄 تست edits1:")
    test_word = "teacher"
    for _ in range(3):
        edited = edits1(test_word)
        print(f"   '{test_word}' -> '{edited}'")
    
    print("=" * 60)
```

```python
"""
مرحله 1: ترکیب دو فایل و حذف کلمات چندکلمه‌ای
- ورودی 1: Persian-Words-Database/distinct_words.txt (50000 خط اول)
- ورودی 2: corpora_farsi/farhang-farsi.tr (کل فایل)
- خروجی: corpora_farsi/corpora-farsi.tr 
- قوانین:
  ✅ نیم‌فاصله (U+200C) حفظ می‌شود
  ❌ کلمات با فاصله معمولی (U+0020) حذف می‌شوند
  ✅ حذف تکراری‌ها
"""

import os

# ===== تنظیمات =====
INPUT_FILE_1 = '/home/GAN_Writing_Farsi/Persian-Words-Database/distinct_words.txt'
INPUT_FILE_2 = '/home/GAN_Writing_Farsi/corpora_farsi/farhang-farsi.tr'
OUTPUT_FILE = '/home/GAN_Writing_Farsi/corpora_farsi/corpora-farsi.tr'
MAX_LINES_FILE1 = 50000

SPACE = ' '  # U+0020
ZWNJ = '\u200c'  # نیم‌فاصله

print("="*70)
print("🔧 مرحله 1: ترکیب و حذف کلمات چندکلمه‌ای")
print("="*70)
print(f"📥 ورودی 1: {INPUT_FILE_1} (50000 خط اول)")
print(f"📥 ورودی 2: {INPUT_FILE_2} (کل)")
print(f"📤 خروجی: {OUTPUT_FILE}\n")

# ===== تابع پردازش فایل =====
def process_file(filepath, max_lines=None):
    """
    خواندن و فیلتر کردن یک فایل
    """
    if not os.path.exists(filepath):
        print(f"⚠️ فایل {filepath} یافت نشد - رد می‌شود")
        return set(), {'total': 0, 'empty': 0, 'has_space': 0, 'has_zwnj': 0, 'valid': 0}
    
    print(f"📖 در حال خواندن: {filepath}")
    
    with open(filepath, 'r', encoding='utf-8') as f:
        lines = f.readlines()
    
    total_read = len(lines)
    
    if max_lines:
        lines = lines[:max_lines]
        print(f"   ✅ کل خطوط: {total_read:,} → محدود به: {len(lines):,}")
    else:
        print(f"   ✅ کل خطوط: {total_read:,}")
    
    valid_words = set()
    stats = {
        'total': len(lines),
        'empty': 0,
        'has_space': 0,
        'has_zwnj': 0,
        'valid': 0
    }
    
    for i, line in enumerate(lines, 1):
        word = line.strip()
        
        # ❌ خط خالی
        if not word:
            stats['empty'] += 1
            continue
        
        # ❌ دارای فاصله معمولی
        if SPACE in word:
            stats['has_space'] += 1
            if i <= 10:  # نمایش 10 نمونه اول
                print(f"      {i:6d}. ❌ '{word}' → دارای فاصله")
            continue
        
        # ✅ معتبر
        if ZWNJ in word:
            stats['has_zwnj'] += 1
        
        valid_words.add(word)
        stats['valid'] += 1
        
        # نمایش نمونه
        if i <= 10:
            has_zwnj_mark = '🔸' if ZWNJ in word else '  '
            print(f"      {i:6d}. ✅ '{word}' {has_zwnj_mark}")
    
    return valid_words, stats

# ===== پردازش فایل اول =====
print("\n" + "="*70)
print("📁 پردازش فایل 1: distinct_words.txt")
print("="*70)
words_set_1, stats_1 = process_file(INPUT_FILE_1, MAX_LINES_FILE1)

# ===== پردازش فایل دوم =====
print("\n" + "="*70)
print("📁 پردازش فایل 2: corpora-farsi.tr (موجود)")
print("="*70)
words_set_2, stats_2 = process_file(INPUT_FILE_2)

# ===== ترکیب =====
print("\n" + "="*70)
print("🔗 ترکیب و حذف تکراری‌ها")
print("="*70)
print(f"   📦 فایل 1: {len(words_set_1):,} کلمه یونیک")
print(f"   📦 فایل 2: {len(words_set_2):,} کلمه یونیک")

all_words = words_set_1 | words_set_2  # Union
duplicates = len(words_set_1) + len(words_set_2) - len(all_words)

print(f"   ➕ مجموع:  {len(words_set_1) + len(words_set_2):,} کلمه")
print(f"   ♻️ تکراری: {duplicates:,} کلمه")
print(f"   ✅ یونیک:   {len(all_words):,} کلمه")

# ===== مرتب‌سازی =====
sorted_words = sorted(all_words)
print(f"\n   🔤 مرتب‌سازی الفبایی انجام شد")

# ===== ذخیره =====
print("\n💾 در حال ذخیره...")

# ایجاد پوشه
output_dir = os.path.dirname(OUTPUT_FILE)
if not os.path.exists(output_dir):
    os.makedirs(output_dir)
    print(f"   ✅ پوشه {output_dir} ایجاد شد")

with open(OUTPUT_FILE, 'w', encoding='utf-8') as f:
    f.write('\n'.join(sorted_words))

print(f"   ✅ {len(sorted_words):,} کلمه ذخیره شد")

# ===== آمار کلی =====
total_input = stats_1['total'] + stats_2['total']
total_empty = stats_1['empty'] + stats_2['empty']
total_space = stats_1['has_space'] + stats_2['has_space']
total_zwnj = stats_1['has_zwnj'] + stats_2['has_zwnj']

words_with_zwnj = [w for w in all_words if ZWNJ in w]

print("\n" + "="*70)
print("📊 آمار نهایی:")
print("="*70)
print(f"   📥 کل خطوط پردازش شده:     {total_input:,}")
print(f"      ├─ فایل 1:                {stats_1['total']:,}")
print(f"      └─ فایل 2:                {stats_2['total']:,}")
print(f"   ❌ خط خالی:                  {total_empty:,}")
print(f"   ❌ دارای فاصله معمولی:       {total_space:,} ({100*total_space/total_input if total_input else 0:.1f}%)")
print(f"   ✅ کلمات یونیک نهایی:       {len(all_words):,}")
print(f"   🔸 با نیم‌فاصله:            {len(words_with_zwnj):,} ({100*len(words_with_zwnj)/len(all_words):.1f}%)")
print("="*70)
print(f"📁 فایل خروجی: {OUTPUT_FILE}")
print("="*70)
print("✅ مرحله 1 تمام شد!\n")

# ===== نمایش نمونه =====
print("🔍 نمونه کلمات نهایی (10 مورد اول):")
print("="*70)
for i, word in enumerate(sorted_words[:10], 1):
    has_zwnj_mark = '🔸' if ZWNJ in word else '  '
    print(f"   {i:2d}. '{word}' {has_zwnj_mark}")
print("="*70)
```

```python
"""
مرحله 2: فیلتر کلمات بر اساس الفبای مجاز و طول مرئی
- ورودی: corpora_farsi/corpora-farsi.tr
- خروجی: corpora_farsi/corpora-farsi-myChars.tr
- الفبای مجاز: 30 کاراکتر (نیم‌فاصله + آابتثجحخدذرزسشصضعفقلمنهوپچکگی)
- حداکثر طول مرئی: 10 حرف (بدون نیم‌فاصله)
"""

import os

# ===== تنظیمات =====
INPUT_FILE = '/home/GAN_Writing_Farsi/corpora_farsi/corpora-farsi.tr'
OUTPUT_FILE = '/home/GAN_Writing_Farsi/corpora_farsi/corpora-farsi-myChars.tr'

# ⚠️ فقط نیم‌فاصله - بدون فاصله معمولی
ALLOWED_CHARS = set('آابتثجحخدذرزسشصضعفقلمنهوپچکگی\u200c')  # 30 کاراکتر
MAX_CHARS = 10  # حداکثر طول مرئی (بدون نیم‌فاصله)

def count_visible_chars(word):
    """شمارش حروف بدون نیم‌فاصله"""
    return len(word.replace('\u200c', ''))

print("="*70)
print("🔧 مرحله 2: فیلتر بر اساس الفبای مجاز و طول مرئی")
print("="*70)
print(f"📥 ورودی: {INPUT_FILE}")
print(f"📤 خروجی: {OUTPUT_FILE}")
print(f"🔤 الفبای مجاز: {len(ALLOWED_CHARS)} کاراکتر")
print(f"📏 حداکثر طول مرئی: {MAX_CHARS} حرف (بدون نیم‌فاصله)")

# نمایش کاراکترها
chars_display = list(ALLOWED_CHARS)
chars_display.sort()
zwnj_index = chars_display.index('\u200c') if '\u200c' in chars_display else -1
if zwnj_index >= 0:
    chars_display[zwnj_index] = 'نیم‌فاصله'
print(f"   → {' '.join(chars_display)}\n")

# ===== بررسی وجود فایل ورودی =====
if not os.path.exists(INPUT_FILE):
    print(f"❌ خطا: فایل {INPUT_FILE} پیدا نشد!")
    print("⚠️ لطفاً ابتدا step1_remove_multiword.py را اجرا کنید")
    exit(1)

# ===== خواندن فایل ورودی =====
print("📖 در حال خواندن فایل ورودی...")
with open(INPUT_FILE, 'r', encoding='utf-8') as f:
    input_words = f.read().strip().split('\n')

print(f"✅ تعداد کلمات ورودی: {len(input_words):,}\n")

# ===== فیلتر کردن =====
print("🔍 در حال فیلتر کردن...\n")

valid_words = []
invalid_samples = []
stats = {
    'total': len(input_words),
    'invalid_chars': 0,
    'too_long': 0,
    'too_short': 0,  # ✅ اضافه شد
    'valid': 0
}

for i, word in enumerate(input_words, 1):
    word = word.strip()
    
    if not word:
        continue
    
    # ✅ چک طول مرئی (اول این را چک کنیم)
    visible_len = count_visible_chars(word)
    
    # حذف کلمات خیلی کوتاه (فقط نیم‌فاصله)
    if visible_len == 0:
        stats['too_short'] += 1
        if i <= 20 or (i % 5000 == 0):
            print(f"   {i:6d}. ❌ '{word}' → طول مرئی=0 (فقط نیم‌فاصله)")
        continue
    
    if visible_len > MAX_CHARS:
        stats['too_long'] += 1
        
        if len(invalid_samples) < 20:
            invalid_samples.append({
                'index': i,
                'word': word,
                'invalid': [f'طول_مرئی={visible_len}>{MAX_CHARS}']
            })
        
        if i <= 20 or (i % 5000 == 0):
            print(f"   {i:6d}. ❌ '{word}' → طول مرئی={visible_len} (کل={len(word)})")
        
        continue
    
    # چک کاراکترهای نامعتبر
    invalid_chars = []
    for char in word:
        if char not in ALLOWED_CHARS:
            char_display = f"{char}(U+{ord(char):04X})"
            # برای فاصله معمولی پیام خاص
            if char == ' ':
                char_display = "SPACE(U+0020)"
            invalid_chars.append(char_display)
    
    if invalid_chars:
        stats['invalid_chars'] += 1
        
        if len(invalid_samples) < 40:  # بیشتر کنیم برای دو نوع خطا
            invalid_samples.append({
                'index': i,
                'word': word,
                'invalid': invalid_chars
            })
        
        if i <= 20 or (i % 5000 == 0):
            print(f"   {i:6d}. ❌ '{word}' → {', '.join(invalid_chars[:3])}")
        
        continue
    
    # ✅ معتبر
    valid_words.append(word)
    stats['valid'] += 1
    
    if i <= 20 or (i % 5000 == 0):
        has_zwnj = '🔸' if '\u200c' in word else '  '
        print(f"   {i:6d}. ✅ '{word}' {has_zwnj} (طول مرئی={visible_len})")

# ===== ذخیره =====
print("\n💾 در حال ذخیره...")
with open(OUTPUT_FILE, 'w', encoding='utf-8') as f:
    f.write('\n'.join(valid_words))

print(f"✅ {len(valid_words):,} کلمه ذخیره شد\n")

# ===== آمار نهایی =====
lengths_total = [len(w) for w in valid_words]
lengths_visible = [count_visible_chars(w) for w in valid_words]
with_zwnj = [w for w in valid_words if '\u200c' in w]

print("="*70)
print("📊 آمار نهایی:")
print("="*70)
print(f"   📥 کل کلمات ورودی:         {stats['total']:,}")
print(f"   ❌ کاراکتر نامعتبر:         {stats['invalid_chars']:,} ({100*stats['invalid_chars']/stats['total']:.1f}%)")
print(f"   ❌ طول بیش از حد:           {stats['too_long']:,} ({100*stats['too_long']/stats['total']:.1f}%)")
print(f"   ❌ طول خیلی کم (=0):        {stats['too_short']:,} ({100*stats['too_short']/stats['total']:.2f}%)")
print(f"   ✅ معتبر (الفبای مجاز):    {stats['valid']:,} ({100*stats['valid']/stats['total']:.1f}%)")
print(f"   🔸 با نیم‌فاصله:            {len(with_zwnj):,} ({100*len(with_zwnj)/len(valid_words):.1f}%)")
print()
print(f"   📏 طول کل (با نیم‌فاصله):")
if lengths_total:
    print(f"      • کوتاه‌ترین:           {min(lengths_total)} حرف")
    print(f"      • بلندترین:             {max(lengths_total)} حرف")
    print(f"      • متوسط:                {sum(lengths_total)/len(lengths_total):.2f} حرف")
print()
print(f"   📐 طول مرئی (بدون نیم‌فاصله):")
if lengths_visible:
    print(f"      • کوتاه‌ترین:           {min(lengths_visible)} حرف")
    print(f"      • بلندترین:             {max(lengths_visible)} حرف")
    print(f"      • متوسط:                {sum(lengths_visible)/len(lengths_visible):.2f} حرف")
print("="*70)
print(f"📁 فایل خروجی: {OUTPUT_FILE}")
print("="*70)
print("✅ مرحله 2 تمام شد!\n")

# نمایش نمونه کلمات حذف شده
if invalid_samples:
    print("\n🔍 نمونه کلمات حذف شده (20 مورد اول):")
    print("="*70)
    
    # جدا کردن دو نوع خطا
    too_long_samples = [s for s in invalid_samples if any('طول_مرئی' in str(inv) for inv in s['invalid'])]
    invalid_char_samples = [s for s in invalid_samples if s not in too_long_samples]
    
    if too_long_samples:
        print("\n   📏 طول بیش از حد:")
        for sample in too_long_samples[:10]:
            vis_len = count_visible_chars(sample['word'])
            print(f"      {sample['index']:6d}. '{sample['word']}'")
            print(f"              → طول کل={len(sample['word'])}, مرئی={vis_len} > {MAX_CHARS}")
    
    if invalid_char_samples:
        print("\n   🔤 کاراکتر نامعتبر:")
        for sample in invalid_char_samples[:10]:
            print(f"      {sample['index']:6d}. '{sample['word']}'")
            print(f"              → {', '.join(sample['invalid'][:5])}")
    
    print("="*70)
```

```python
import os
import torch.utils.data as D
import random
import cv2
import numpy as np
from pairs_idx_wid_ABAN import wid2label_tr, wid2label_te

CREATE_PAIRS = False

IMG_HEIGHT = 64
IMG_WIDTH = 216
MAX_CHARS = 10
NUM_CHANNEL = 15
# NUM_CHANNEL = 50
EXTRA_CHANNEL = NUM_CHANNEL + 1
NUM_WRITERS = 500  # ABAN: 350 train + 150 test
NORMAL = True
OUTPUT_MAX_LEN = MAX_CHARS + 2  # <GO>+groundtruth+<END>

"""مسیر تصاویر کلمات ABAN"""
img_base = "./datasets/aban/words"
text_corpus = "./corpora_farsi/corpora-farsi-myChars.tr"

with open(text_corpus, "r", encoding='utf-8') as _f:
    text_corpus = _f.read().strip().split('\n')

src = "Groundtruth_farsi/gan.aban.tr_va.gt.filter27"
tar = "Groundtruth_farsi/gan.aban.test.gt.filter27"


def labelDictionary():
    """
    ساخت دیکشنری الفبای فارسی
    30 کاراکتر: آابتثجحخدذرزسشصضعفقلمنهوپچکگی + نیم‌فاصله
    """
    ZWNJ = '\u200c'  # Zero-Width Non-Joiner
    labels = list('آابتثجحخدذرزسشصضعفقلمنهوپچکگی' + ZWNJ)
    letter2index = {label: n for n, label in enumerate(labels)}
    index2letter = {v: k for k, v in letter2index.items()}
    return len(labels), letter2index, index2letter


num_classes, letter2index, index2letter = labelDictionary()
tokens = {"GO_TOKEN": 0, "END_TOKEN": 1, "PAD_TOKEN": 2}
num_tokens = len(tokens.keys())
vocab_size = num_classes + num_tokens


def edits1(word, min_len=2, max_len=MAX_CHARS):
    """
    ایجاد یک ویرایش (edit distance 1) از کلمه فارسی
    عملیات: حذف، جابجایی، جایگزینی، درج
    توجه: طول بر اساس حروف مرئی (بدون نیم‌فاصله) محاسبه می‌شود
    """
    ZWNJ = '\u200c'
    letters = list('آابتثجحخدذرزسشصضعفقلمنهوپچکگی' + ZWNJ)
    
    splits = [(word[:i], word[i:]) for i in range(len(word) + 1)]
    deletes = [L + R[1:] for L, R in splits if R]
    transposes = [L + R[1] + R[0] + R[2:] for L, R in splits if len(R) > 1]
    replaces = [L + c + R[1:] for L, R in splits if R for c in letters]
    inserts = [L + c + R for L, R in splits for c in letters]
    
    # محاسبه طول مرئی (بدون نیم‌فاصله)
    visible_len = len([c for c in word if c != ZWNJ])
    
    if visible_len <= min_len:
        # کلمه کوتاه: فقط جابجایی، جایگزینی، درج
        candidates = list(set(transposes + replaces + inserts))
    elif visible_len >= max_len:
        # کلمه بلند: فقط حذف، جابجایی، جایگزینی
        candidates = list(set(deletes + transposes + replaces))
    else:
        # کلمه متوسط: همه عملیات
        candidates = list(set(deletes + transposes + replaces + inserts))
    
    if not candidates:
        return word
    return random.choice(candidates)


class ABAN_words(D.Dataset):
    def __init__(self, data_dict, oov):
        self.data_dict = data_dict
        self.oov = oov
        self.output_max_len = OUTPUT_MAX_LEN

    def new_ed1(self, word_ori):
        """
        ایجاد یک کلمه جدید با edit distance 1 از کلمه اصلی
        word_ori مثال: [0, 15, 27, 13, 32, 31, 1, 2, 2, 2]
        """
        word = word_ori.copy()
        start = word.index(tokens["GO_TOKEN"])
        fin = word.index(tokens["END_TOKEN"])
        word = "".join([index2letter[i - num_tokens] for i in word[start + 1 : fin]])
        new_word = edits1(word)
        label = np.array(self.label_padding(new_word, num_tokens))
        return label

    def __getitem__(self, wid_idx_num):
        words = self.data_dict[wid_idx_num]
        """shuffle images"""
        np.random.shuffle(words)

        wids = list()
        idxs = list()
        imgs = list()
        img_widths = list()
        labels = list()

        for word in words:
            wid, idx = word[0].split(",")
            img, img_width = self.read_image_single(idx)
            label = self.label_padding(" ".join(word[1:]), num_tokens)
            wids.append(wid)
            idxs.append(idx)
            imgs.append(img)
            img_widths.append(img_width)
            labels.append(label)

        if len(list(set(wids))) != 1:
            print("Error! writer id differs")
            exit()

        final_wid = wid_idx_num
        num_imgs = len(imgs)
        if num_imgs >= EXTRA_CHANNEL:
            final_img = np.stack(imgs[:EXTRA_CHANNEL], axis=0)  # 64, h, w
            final_idx = idxs[:EXTRA_CHANNEL]
            final_img_width = img_widths[:EXTRA_CHANNEL]
            final_label = labels[:EXTRA_CHANNEL]
        else:
            final_idx = idxs
            final_img = imgs
            final_img_width = img_widths
            final_label = labels

            while len(final_img) < EXTRA_CHANNEL:
                num_cp = EXTRA_CHANNEL - len(final_img)
                final_idx = final_idx + idxs[:num_cp]
                final_img = final_img + imgs[:num_cp]
                final_img_width = final_img_width + img_widths[:num_cp]
                final_label = final_label + labels[:num_cp]
            final_img = np.stack(final_img, axis=0)

        _id = np.random.randint(EXTRA_CHANNEL)
        img_xt = final_img[_id : _id + 1]
        if self.oov:
            label_xt = np.random.choice(text_corpus)
            label_xt = np.array(self.label_padding(label_xt, num_tokens))
            label_xt_swap = np.random.choice(text_corpus)
            label_xt_swap = np.array(self.label_padding(label_xt_swap, num_tokens))
        else:
            label_xt = final_label[_id]
            label_xt_swap = self.new_ed1(label_xt)

        final_idx = np.delete(final_idx, _id, axis=0)
        final_img = np.delete(final_img, _id, axis=0)
        final_img_width = np.delete(final_img_width, _id, axis=0)
        final_label = np.delete(final_label, _id, axis=0)

        return (
            "src",
            final_wid,
            final_idx,
            final_img,
            final_img_width,
            final_label,
            img_xt,
            label_xt,
            label_xt_swap,
        )

    def __len__(self):
        return len(self.data_dict)

    def read_image_single(self, file_name):
        """
        خواندن تصویر کلمه فارسی
        فرمت: .png (مشابه IAM)
        """
        url = os.path.join(img_base, file_name + ".png")
        img = cv2.imread(url, 0)

        if img is None and os.path.exists(url):
            # image is present but corrupted
            return np.zeros((IMG_HEIGHT, IMG_WIDTH)), 0

        rate = float(IMG_HEIGHT) / img.shape[0]
        img = cv2.resize(
            img,
            (int(img.shape[1] * rate) + 1, IMG_HEIGHT),
            interpolation=cv2.INTER_CUBIC,
        )
        img = img / 255.0  # 0-255 -> 0-1

        img = 1.0 - img
        img_width = img.shape[-1]

        if img_width > IMG_WIDTH:
            outImg = img[:, :IMG_WIDTH]
            img_width = IMG_WIDTH
        else:
            outImg = np.zeros((IMG_HEIGHT, IMG_WIDTH), dtype="float32")
            outImg[:, :img_width] = img
        outImg = outImg.astype("float32")

        mean = 0.5
        std = 0.5
        outImgFinal = (outImg - mean) / std
        return outImgFinal, img_width

    def label_padding(self, labels, num_tokens):
        """
        تبدیل رشته به لیست اندیس‌ها با padding
        مثال: "س ل ا م" -> [0, 15, 27, 13, 32, 1, 2, 2, 2, 2, 2, 2]
        """
        new_label_len = []
        ll = [letter2index[i] for i in labels]
        new_label_len.append(len(ll) + 2)
        ll = np.array(ll) + num_tokens
        ll = list(ll)
        ll = [tokens["GO_TOKEN"]] + ll + [tokens["END_TOKEN"]]
        num = self.output_max_len - len(ll)
        if not num == 0:
            ll.extend([tokens["PAD_TOKEN"]] * num)  # replace PAD_TOKEN
        return ll


def loadData(oov):
    """
    بارگذاری داده‌های آموزش و تست ABAN
    """
    gt_tr = src
    gt_te = tar

    with open(gt_tr, "r", encoding='utf-8') as f_tr:
        data_tr = f_tr.readlines()
        data_tr = [i.strip().split(" ") for i in data_tr]
        tr_dict = dict()
        for i in data_tr:
            wid = i[0].split(",")[0]
            if wid not in tr_dict.keys():
                tr_dict[wid] = [i]
            else:
                tr_dict[wid].append(i)
        new_tr_dict = dict()
        if CREATE_PAIRS:
            create_pairs(tr_dict)
        for k in tr_dict.keys():
            new_tr_dict[wid2label_tr[k]] = tr_dict[k]

    with open(gt_te, "r", encoding='utf-8') as f_te:
        data_te = f_te.readlines()
        data_te = [i.strip().split(" ") for i in data_te]
        te_dict = dict()
        for i in data_te:
            wid = i[0].split(",")[0]
            if wid not in te_dict.keys():
                te_dict[wid] = [i]
            else:
                te_dict[wid].append(i)
        new_te_dict = dict()
        if CREATE_PAIRS:
            create_pairs(te_dict)
        for k in te_dict.keys():
            new_te_dict[wid2label_te[k]] = te_dict[k]

    data_train = ABAN_words(new_tr_dict, oov)
    data_test = ABAN_words(new_te_dict, oov)
    return data_train, data_test


def create_pairs(ddict):
    """
    ایجاد جفت‌های (label, writer_id)
    """
    num = len(ddict.keys())
    label2wid = list(zip(range(num), ddict.keys()))
    print(label2wid)


if __name__ == "__main__":
    # تست سریع
    ZWNJ = '\u200c'
    print("=" * 60)
    print("📚 تست load_data_farsi.py")
    print("=" * 60)
    print(f"✓ تعداد کلاس‌ها (حروف فارسی): {num_classes}")
    print(f"✓ vocab_size (حروف + توکن‌ها): {vocab_size}")
    print(f"✓ NUM_WRITERS: {NUM_WRITERS}")
    print(f"✓ OUTPUT_MAX_LEN: {OUTPUT_MAX_LEN}")
    print("\n📖 نمونه الفبا:")
    for i, (char, idx) in enumerate(list(letter2index.items())[:5]):
        print(f"   '{char}' -> {idx}")
    print(f"   ...")
    zwnj_idx = letter2index[ZWNJ]
    print(f"   نیم‌فاصله (U+200C) -> {zwnj_idx}")
    
    print("\n🔄 تست edits1:")
    test_word = "سلام"
    for _ in range(3):
        edited = edits1(test_word)
        print(f"   '{test_word}' -> '{edited}'")
    
    print("=" * 60)
```

```python
import os
import torch.utils.data as D
import random
import string
import cv2
import numpy as np
from pairs_idx_wid_iam import wid2label_tr, wid2label_te

CREATE_PAIRS = False

IMG_HEIGHT = 64
IMG_WIDTH = 216
MAX_CHARS = 10
NUM_CHANNEL = 15
# NUM_CHANNEL = 50
EXTRA_CHANNEL = NUM_CHANNEL + 1
NUM_WRITERS = 500  # iam
NORMAL = True
OUTPUT_MAX_LEN = MAX_CHARS + 2  # <GO>+groundtruth+<END>

"""The folder of IAM word images, please change to your own one before run it!!"""
img_base = "./datasets/iam/words"
text_corpus = "corpora_english/brown-azAZ.tr"

with open(text_corpus, "r") as _f:
    text_corpus = _f.read().split()

src = "Groundtruth/gan.iam.tr_va.gt.filter27"
tar = "Groundtruth/gan.iam.test.gt.filter27"


def labelDictionary():
    labels = list(string.ascii_lowercase + string.ascii_uppercase)
    letter2index = {label: n for n, label in enumerate(labels)}
    index2letter = {v: k for k, v in letter2index.items()}
    return len(labels), letter2index, index2letter


num_classes, letter2index, index2letter = labelDictionary()
tokens = {"GO_TOKEN": 0, "END_TOKEN": 1, "PAD_TOKEN": 2}
num_tokens = len(tokens.keys())
vocab_size = num_classes + num_tokens


def edits1(word, min_len=2, max_len=MAX_CHARS):
    "All edits that are one edit away from `word`."
    letters = list(string.ascii_lowercase)
    splits = [(word[:i], word[i:]) for i in range(len(word) + 1)]
    deletes = [L + R[1:] for L, R in splits if R]
    transposes = [L + R[1] + R[0] + R[2:] for L, R in splits if len(R) > 1]
    replaces = [L + c + R[1:] for L, R in splits if R for c in letters]
    inserts = [L + c + R for L, R in splits for c in letters]
    if len(word) <= min_len:
        return random.choice(list(set(transposes + replaces + inserts)))
    elif len(word) >= max_len:
        return random.choice(list(set(deletes + transposes + replaces)))
    else:
        return random.choice(list(set(deletes + transposes + replaces + inserts)))


class IAM_words(D.Dataset):
    def __init__(self, data_dict, oov):
        self.data_dict = data_dict
        self.oov = oov
        self.output_max_len = OUTPUT_MAX_LEN

    # word [0, 15, 27, 13, 32, 31, 1, 2, 2, 2]
    def new_ed1(self, word_ori):
        word = word_ori.copy()
        start = word.index(tokens["GO_TOKEN"])
        fin = word.index(tokens["END_TOKEN"])
        word = "".join([index2letter[i - num_tokens] for i in word[start + 1 : fin]])
        new_word = edits1(word)
        label = np.array(self.label_padding(new_word, num_tokens))
        return label

    def __getitem__(self, wid_idx_num):
        # print("wid_idx_num: ", wid_idx_num)
        words = self.data_dict[wid_idx_num]
        """shuffle images"""
        np.random.shuffle(words)

        wids = list()
        idxs = list()
        imgs = list()
        img_widths = list()
        labels = list()

        for word in words:
            wid, idx = word[0].split(",")
            img, img_width = self.read_image_single(idx)
            label = self.label_padding(" ".join(word[1:]), num_tokens)
            wids.append(wid)
            idxs.append(idx)
            imgs.append(img)
            img_widths.append(img_width)
            labels.append(label)

        if len(list(set(wids))) != 1:
            print("Error! writer id differs")
            exit()

        final_wid = wid_idx_num
        num_imgs = len(imgs)
        if num_imgs >= EXTRA_CHANNEL:
            final_img = np.stack(imgs[:EXTRA_CHANNEL], axis=0)  # 64, h, w
            final_idx = idxs[:EXTRA_CHANNEL]
            final_img_width = img_widths[:EXTRA_CHANNEL]
            final_label = labels[:EXTRA_CHANNEL]
        else:
            final_idx = idxs
            final_img = imgs
            final_img_width = img_widths
            final_label = labels

            while len(final_img) < EXTRA_CHANNEL:
                num_cp = EXTRA_CHANNEL - len(final_img)
                final_idx = final_idx + idxs[:num_cp]
                final_img = final_img + imgs[:num_cp]
                final_img_width = final_img_width + img_widths[:num_cp]
                final_label = final_label + labels[:num_cp]
            final_img = np.stack(final_img, axis=0)

        _id = np.random.randint(EXTRA_CHANNEL)
        img_xt = final_img[_id : _id + 1]
        if self.oov:
            label_xt = np.random.choice(text_corpus)
            label_xt = np.array(self.label_padding(label_xt, num_tokens))
            label_xt_swap = np.random.choice(text_corpus)
            label_xt_swap = np.array(self.label_padding(label_xt_swap, num_tokens))
        else:
            label_xt = final_label[_id]
            label_xt_swap = self.new_ed1(label_xt)

        final_idx = np.delete(final_idx, _id, axis=0)
        final_img = np.delete(final_img, _id, axis=0)
        final_img_width = np.delete(final_img_width, _id, axis=0)
        final_label = np.delete(final_label, _id, axis=0)

        return (
            "src",
            final_wid,
            final_idx,
            final_img,
            final_img_width,
            final_label,
            img_xt,
            label_xt,
            label_xt_swap,
        )

    def __len__(self):
        return len(self.data_dict)

    def read_image_single(self, file_name):
        url = os.path.join(img_base, file_name + ".png")
        img = cv2.imread(url, 0)

        if img is None and os.path.exists(url):
            # image is present but corrupted
            return np.zeros((IMG_HEIGHT, IMG_WIDTH)), 0

        rate = float(IMG_HEIGHT) / img.shape[0]
        img = cv2.resize(
            img,
            (int(img.shape[1] * rate) + 1, IMG_HEIGHT),
            interpolation=cv2.INTER_CUBIC,
        )
        img = img / 255.0  # 0-255 -> 0-1

        img = 1.0 - img
        img_width = img.shape[-1]

        if img_width > IMG_WIDTH:
            outImg = img[:, :IMG_WIDTH]
            img_width = IMG_WIDTH
        else:
            outImg = np.zeros((IMG_HEIGHT, IMG_WIDTH), dtype="float32")
            outImg[:, :img_width] = img
        outImg = outImg.astype("float32")

        mean = 0.5
        std = 0.5
        outImgFinal = (outImg - mean) / std
        return outImgFinal, img_width

    def label_padding(self, labels, num_tokens):
        new_label_len = []
        ll = [letter2index[i] for i in labels]
        new_label_len.append(len(ll) + 2)
        ll = np.array(ll) + num_tokens
        ll = list(ll)
        ll = [tokens["GO_TOKEN"]] + ll + [tokens["END_TOKEN"]]
        num = self.output_max_len - len(ll)
        if not num == 0:
            ll.extend([tokens["PAD_TOKEN"]] * num)  # replace PAD_TOKEN
        return ll


def loadData(oov):
    gt_tr = src
    gt_te = tar

    with open(gt_tr, "r") as f_tr:
        data_tr = f_tr.readlines()
        data_tr = [i.strip().split(" ") for i in data_tr]
        tr_dict = dict()
        for i in data_tr:
            wid = i[0].split(",")[0]
            if wid not in tr_dict.keys():
                tr_dict[wid] = [i]
            else:
                tr_dict[wid].append(i)
        new_tr_dict = dict()
        if CREATE_PAIRS:
            create_pairs(tr_dict)
        for k in tr_dict.keys():
            new_tr_dict[wid2label_tr[k]] = tr_dict[k]

    with open(gt_te, "r") as f_te:
        data_te = f_te.readlines()
        data_te = [i.strip().split(" ") for i in data_te]
        te_dict = dict()
        for i in data_te:
            wid = i[0].split(",")[0]
            if wid not in te_dict.keys():
                te_dict[wid] = [i]
            else:
                te_dict[wid].append(i)
        new_te_dict = dict()
        if CREATE_PAIRS:
            create_pairs(te_dict)
        for k in te_dict.keys():
            new_te_dict[wid2label_te[k]] = te_dict[k]
    
    # print(new_te_dict)
    data_train = IAM_words(new_tr_dict, oov)
    data_test = IAM_words(new_te_dict, oov)
    return data_train, data_test


def create_pairs(ddict):
    num = len(ddict.keys())
    label2wid = list(zip(range(num), ddict.keys()))
    print(label2wid)


if __name__ == "__main__":
    # تست سریع
    print("=" * 60)
    print("📚 تست load_data.py")
    print("=" * 60)
    print(f"✓ تعداد کلاس‌ها (حروف انگلیسی): {num_classes}")
    print(f"✓ vocab_size (حروف + توکن‌ها): {vocab_size}")
    print(f"✓ NUM_WRITERS: {NUM_WRITERS}")
    print(f"✓ OUTPUT_MAX_LEN: {OUTPUT_MAX_LEN}")
    print("\n📖 نمونه الفبا:")
    for i, (char, idx) in enumerate(list(letter2index.items())[:5]):
        print(f"   '{char}' -> {idx}")
    print(f"   ...")
    # print(f"   نیم‌فاصله (\\u200c) -> {letter2index['\u200c']}")
    
    print("\n🔄 تست edits1:")
    test_word = "teacher"
    for _ in range(3):
        edited = edits1(test_word)
        print(f"   '{test_word}' -> '{edited}'")
    
    print("=" * 60)
```
