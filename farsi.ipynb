# Create Groundtruth_farsi


```python
import os
import random

# Path to the image folder
img_base = "./datasets/TC_Words/words"

# Path to the output folder
output_dir = "./Groundtruth_farsi/"
os.makedirs(output_dir, exist_ok=True)

# Output files
train_file_path = os.path.join(output_dir, "gan.persian.tr_va.gt.filter27")
test_file_path = os.path.join(output_dir, "gan.persian.test.gt.filter27")

# List for temporarily storing lines
data_dict = {}

# Collect data from folders
for word in os.listdir(img_base):
    word_path = os.path.join(img_base, word)
    if os.path.isdir(word_path):
        for image in os.listdir(word_path):
            if image.endswith(".tif"):  # Ensure the file is an image
                # Extract author ID from the last three digits of the file ID
                author_id = image.split("_")[0][2:]
                # Construct the appropriate line for the output files
                line = f"{author_id},{word}/{image.split('.')[0]} {word}\n"
                if author_id not in data_dict:
                    data_dict[author_id] = []
                data_dict[author_id].append((image.split(".")[0], line))

# Split authors into two sets
author_ids = list(data_dict.keys())
random.shuffle(author_ids)
split_index = int(0.7 * len(author_ids))
train_authors = set(author_ids[:split_index])
test_authors = set(author_ids[split_index:])

# Create pairs_tr and pairs_te lists
pairs_tr = []
pairs_te = []

# Write data to output files and create the pairs lists
with open(train_file_path, "w", encoding="utf-8") as train_file, open(
    test_file_path, "w", encoding="utf-8"
) as test_file:
    for author_id, lines in data_dict.items():
        if author_id in train_authors:
            for wid, line in lines:
                train_file.write(line)
            pairs_tr.append((len(pairs_tr), author_id))
        elif author_id in test_authors:
            for wid, line in lines:
                test_file.write(line)
            pairs_te.append((len(pairs_te), author_id))

print("Files created and sorted successfully!")

# Print the count of data for verification
train_count = sum(
    len(lines) for author_id, lines in data_dict.items() if author_id in train_authors
)
test_count = sum(
    len(lines) for author_id, lines in data_dict.items() if author_id in test_authors
)

print(f"Number of data for training file: {train_count}")
print(f"Number of data for test file: {test_count}")

# Print the created pairs lists for verification
print(f"pairs_tr: {pairs_tr}")
print(f"pairs_te: {pairs_te}")
```

```python
import re


def change_format_groundtruth_file(input_file, output_file):
    """
    Reads a groundtruth file line by line, removes folder names from the records,
    and writes the modified content directly to the output file.

    :param input_file: Path to the input file
    :param output_file: Path to the output file
    """
    with open(input_file, "r", encoding="utf-8") as infile, open(
        output_file, "w", encoding="utf-8"
    ) as outfile:
        for line in infile:
            # Split the line based on commas and spaces (split up to 2 times)
            parts = re.split(r"[ ,]", line.strip(), maxsplit=2)

            # if len(parts) < 3:
            #     continue  # Skip lines that are not properly formatted

            writer_id = parts[0]  # First part is the writer ID
            file_id = parts[1].split("/")[-1]  # Remove the folder name from file ID
            label = parts[-1]  # Last part is the label

            # Write the reformatted line directly to the output file
            outfile.write(f"{writer_id},{file_id} {label}\n")


# Example usage
input_file = "Groundtruth_farsi/gan.persian.tr_va.gt.filter27"
output_file = "Groundtruth_farsi/gan.aban.tr_va.gt.filter27"
change_format_groundtruth_file(input_file, output_file)

input_file = "Groundtruth_farsi/gan.persian.test.gt.filter27"
output_file = "Groundtruth_farsi/gan.aban.test.gt.filter27"
change_format_groundtruth_file(input_file, output_file)
```

```python
def read_file(file_path):
    # Open the file and read all lines, returning them as a set of stripped strings
    with open(file_path, "r", encoding="utf-8") as file:
        return set(line.strip() for line in file)


def count_common_words(file1_path, file2_path):
    # Read words from both files
    words_file1 = read_file(file1_path)
    words_file2 = read_file(file2_path)

    # Find common words
    common_words = words_file1.intersection(words_file2)
    return len(common_words), common_words


file1_path = "/home/ilarima/Payan_Name/GAN_Writing/corpora_english/brown-azAZ.tr"
file2_path = (
    "/home/ilarima/Payan_Name/GAN_Writing/corpora_english/in_vocab.subset.tro.37"
)
file3_path = "/home/ilarima/Payan_Name/GAN_Writing/corpora_english/oov.common_words"

common_count, common_words = count_common_words(file1_path, file2_path)

print(f"Number of common words: {common_count}")
print("Common words:", common_words)

print("=" * 50)
common_count2, common_words2 = count_common_words(file1_path, file3_path)

print(f"Number of common words: {common_count2}")
print("Common words:", common_words2)
```

# Create corpora_farsi


```python
import os
import urllib.request
import pandas as pd


def download_excel(url, download_path):
    """Download the Excel file from the URL and save it to the download_path."""
    if not os.path.exists(download_path):
        print("Downloading file...")
        urllib.request.urlretrieve(url, download_path)
        print("Download complete.")
    else:
        print("File already exists.")


def extract_first_column_and_save(excel_path, output_path):
    """Extract the first column from the Excel file and save it to a .tr file."""
    print("Reading the Excel file...")
    df = pd.read_excel(excel_path, usecols=[0], engine="openpyxl")

    first_column = df.iloc[:, 0].dropna().unique()  # Remove duplicates and NaN values

    print(f"Saving {len(first_column)} unique words to the .tr file...")
    with open(output_path, "w", encoding="utf-8") as f:
        for word in first_column:
            f.write(f"{word}\n")
    print("File saved.")


def cleanup_file(file_path):
    """Remove the specified file."""
    if os.path.exists(file_path):
        os.remove(file_path)
        print(f"File {file_path} has been deleted.")
    else:
        print(f"File {file_path} does not exist.")


if __name__ == "__main__":
    # Define URLs and file paths
    url = "https://bigdata-ir.com/wp-content/uploads/2019/01/%D9%81%D8%A7%DB%8C%D9%84-%D9%81%D8%B1%D9%87%D9%86%DA%AF-%D9%81%D8%A7%D8%B1%D8%B3%DB%8C.xlsx"
    download_path = "./corpora_farsi/farhang-farsi.xlsx"
    output_path = "./corpora_farsi/farhang-farsi.tr"

    # Step 1: Download the Excel file
    download_excel(url, download_path)

    # Step 2: Extract the first column and save to a .tr file
    extract_first_column_and_save(download_path, output_path)

    # Step 3: Clean up the downloaded Excel file
    cleanup_file(download_path)
```

```python
import os
import shutil
from PIL import Image
from tqdm import tqdm


def copyـfolder(src_folder, dest_folder):
    # Check if the source folder exists
    # If the source folder does not exist, print an error and stop
    if not os.path.exists(src_folder):
        print(f"Source folder {src_folder} does not exist.")
        return

    # Copy the source folder to the destination
    # Try to copy the source folder to the destination, and print an error if something goes wrong
    try:
        shutil.copytree(src_folder, dest_folder)
        print(f"Folder {src_folder} copied to {dest_folder}.")
    except Exception as e:
        print(f"Error while copying folder: {e}")
        return


def change_images_format(dest_folder):
    # Count the number of files for tqdm
    # Calculate the total number of files to display the progress bar
    total_files = sum([len(files) for _, _, files in os.walk(dest_folder)])

    # Iterate through the files and folders in the destination folder
    # Display a progress bar while processing files
    with tqdm(total=total_files, desc="Processing Files") as pbar:
        for root, dirs, files in os.walk(dest_folder):
            for file in files:
                file_path = os.path.join(root, file)

                # Remove Thumbs.db files
                # If a Thumbs.db file is found, delete it
                if file.lower() == "thumbs.db":
                    try:
                        os.remove(file_path)
                        # print(f"File {file_path} deleted.")
                    except Exception as e:
                        print(f"Error while deleting file {file_path}: {e}")

                # Convert .tif files to .png
                # If a .tif file is found, convert it to .png
                elif file.lower().endswith(".tif"):
                    try:
                        # Open the .tif image
                        img = Image.open(file_path)

                        # Change the file extension to .png
                        new_file_path = os.path.splitext(file_path)[0] + ".png"

                        # Save the file as .png
                        img.save(new_file_path, "PNG")
                        # print(f"File {file_path} converted to {new_file_path}.")

                        # Remove the original .tif file
                        os.remove(file_path)
                    except Exception as e:
                        print(f"Error while converting file {file_path}: {e}")

                # Update the progress bar
                pbar.update(1)


def delete_folder(folder_path):
    # Check if the folder exists
    # If the folder exists and is a directory, delete it along with all contents
    if os.path.exists(folder_path) and os.path.isdir(folder_path):
        try:
            # Delete the folder and all its contents
            shutil.rmtree(folder_path)
            print(f"Folder {folder_path} deleted successfully.")
        except Exception as e:
            print(f"Error while deleting folder {folder_path}: {e}")
    else:
        print(f"Folder {folder_path} does not exist or is not a directory.")


# Define source and destination folder paths
src_folder = "datasets/TC_Words"  # Path to the source folder
dest_folder = "datasets/ABAN"  # Path to the destination folder
folder_to_delete = "datasets/ABAN/words/Names"  # Folder to be deleted

# Copy the source folder to the destination
copyـfolder(src_folder, dest_folder)

# Delete the specified folder
delete_folder(folder_to_delete)

# Convert image files and remove unnecessary files
change_images_format(dest_folder)
```
